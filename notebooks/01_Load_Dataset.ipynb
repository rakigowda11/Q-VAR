```python
import numpy as np
```

```python
import os

lidar_path = os.path.join(
    nusc.dataroot,
    lidar_data['filename']
)

print(lidar_path)
```

```python
points = np.fromfile(
    lidar_path,
    dtype=np.float32
)
```

```python
print(points.shape)
```

```python
points = points.reshape(-1,5)
```

```python
print(points.shape)
```

```python
print(points[0])
```

```python
print(points[:10])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,10))

plt.scatter(
    points[:,0],
    points[:,1],
    s=0.5
)

plt.xlabel("X")

plt.ylabel("Y")

plt.title("Top View LiDAR Point Cloud")

plt.axis("equal")

plt.show()
```

```python
plt.figure(figsize=(10,10))

plt.scatter(
    points[:,0],
    points[:,1],
    c=points[:,2],
    s=0.5,
    cmap='viridis'
)

plt.colorbar(label="Height (Z)")

plt.xlabel("X")

plt.ylabel("Y")

plt.axis("equal")

plt.title("LiDAR Point Cloud Colored by Height")

plt.show()
```
